# Experiment 13: PAC Learning and VC Dimension

**Aim:** Simulate sample complexity using a toy learning example and understand the PAC model and VC Dimension of linear classifiers.

**Libraries:** NumPy, Matplotlib, Scikit-learn

**Concept:** PAC means Probably Approximately Correct. A learner aims to produce a hypothesis with error at most ε, with confidence at least 1 − δ.

## 1. Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Simulate sample complexity

```python
# Illustrative PAC sample-complexity bound for a finite hypothesis set:
# m >= (1/epsilon) * (ln(|H|) + ln(1/delta))
epsilon = 0.10       # maximum acceptable error
delta = 0.05         # probability that the guarantee fails
hypothesis_count = 10

m = (1 / epsilon) * (
    np.log(hypothesis_count) + np.log(1 / delta)
)

print("Epsilon (error tolerance):", epsilon)
print("Delta (failure probability):", delta)
print("Number of hypotheses:", hypothesis_count)
print("Required sample count (rounded up):", int(np.ceil(m)))
```

## 3. Observe how sample count changes with confidence and accuracy

```python
epsilon_values = [0.30, 0.20, 0.10, 0.05]
delta = 0.05
hypothesis_count = 10

for eps in epsilon_values:
    samples = (1 / eps) * (
        np.log(hypothesis_count) + np.log(1 / delta)
    )
    print("Epsilon =", eps, " Required samples =", int(np.ceil(samples)))
```

## 4. Understand VC Dimension with points in 2D

```python
# For linear classifiers in 2D, the VC dimension is 3.
# Three non-collinear points can be shattered by straight-line classifiers.
points = np.array([[0, 0], [1, 0], [0.5, 1]])

plt.figure(figsize=(6, 5))
plt.scatter(points[:, 0], points[:, 1], s=100)
for i, (x, y) in enumerate(points):
    plt.text(x + 0.03, y + 0.03, "Point " + str(i + 1))
plt.title("Three non-collinear points in 2D")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()

print("For linear classifiers in 2D, VC Dimension = 3.")
```

## 5. Illustrate a labeling that cannot be separated for four points

```python
# Four points at the corners of a square.
# Alternating labels (positive on one diagonal, negative on the other)
# cannot be separated by a single straight line.
square = np.array([[0, 0], [1, 0], [1, 1], [0, 1]])
labels = np.array([1, -1, 1, -1])

plt.figure(figsize=(6, 5))
for point, label in zip(square, labels):
    if label == 1:
        plt.scatter(point[0], point[1], color="blue", marker="o", s=120)
    else:
        plt.scatter(point[0], point[1], color="red", marker="x", s=120)

plt.title("Alternating labels on four square points")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.legend(["Positive class (blue)", "Negative class (red)"])
plt.show()

print("This alternating labeling is not linearly separable.")
print("Therefore, four points in this arrangement cannot be shattered")
print("by 2D linear classifiers. The VC Dimension is 3.")
```

## Observations
1. PAC learning describes learning with a specified accuracy and confidence.
2. ε represents the tolerated error and δ represents the probability of failure.
3. A smaller ε generally requires more training examples.
4. The VC Dimension measures the expressive capacity of a hypothesis class.
5. Linear classifiers in two dimensions have VC Dimension 3.

## Result
The PAC sample-complexity bound was simulated using a finite hypothesis set. The VC Dimension of 2D linear classifiers was studied using toy point arrangements.
